```python
class Node:
    def __init__(self, data):
        self.data = data      # store the data
        self.next = None      # pointer to next node
```

```python
class LinkedList:
    def __init__(self):
        self.head = None     # starting point of the list

    # Add to the end
    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        last = self.head
        while last.next:
            last = last.next
        last.next = new_node

    # Print the list
    def print_list(self):
        current = self.head
        while current:
            print(current.data, end=" → ")
            current = current.next
        print("None")
```

```python
# Create linked list
ll = LinkedList()

ll.append(10)
ll.print_list()
ll.append(20)
ll.print_list()
ll.append(30)
ll.print_list()
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):  # last i elements are already sorted
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
        print(f"current loop {i + 1}: {arr}")
    return arr

print(bubble_sort([5,4,3,2,1]))
```

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i
        print(f"min_index: {min_index}")
        print(f"range(i+1, n): {i+1}, {n}")
        for j in range(i+1, n):
            print(f"compare: {arr[j]} < {arr[min_index]}")
            if arr[j] < arr[min_index]:
                min_index = j
        # Swap if needed
        print(f"compare 2: {min_index} != {i}")
        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            print(".   swapping")
        print(f"current: {arr}")
    return arr

print(selection_sort([5,4,3,2,1]))
```

```python
def selection_sort(arr):
    for i in range(len(arr)):
        current_min_index = i
        for j in range(i + 1, len(arr)):
            if(arr[j] < arr[current_min_index]):
                current_min_index = j
        if(i != current_min_index):
            arr[i], arr[current_min_index] = arr[current_min_index], arr[i]
    return arr

print(selection_sort([5,4,3,2,1]))
```
