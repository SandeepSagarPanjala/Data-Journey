```python
numbers = [123, -456, 7890]

print("Default formatting:")
for n in numbers:
    print('|{:10}|'.format(n))  # right-aligned by default

print("\nLeft aligned with `-10`:")
for n in numbers:
    print('|{:<10}|'.format(n))  # left-aligned
```
